# MemeSense 数据概览
先运行 `python scripts/prepare_data.py scan`。本笔记本只读取中间清单，不修改原图。使用笔记本需另行安装 Jupyter。

```python
import json
from pathlib import Path
from collections import Counter
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
manifest = root / 'data/interim/manifest.jsonl'
if not manifest.exists():
    print('尚无清单，请先运行数据扫描。')
else:
    rows = [json.loads(line) for line in manifest.read_text(encoding='utf-8').splitlines() if line.strip()]
    print('图片数:', len(rows))
    print('唯一文件内容数:', len({row['sha256'] for row in rows}))
    print('扩展名分布:', Counter(Path(row['image_path']).suffix.lower() for row in rows))
```
